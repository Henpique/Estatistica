## Vendo Distribuicoes no Notebook

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
n = 100
media = 3
desvio_padrao = 2
for i in range(3):    
    normal_distribuition = np.random.normal(loc=media, scale=desvio_padrao, size = n)
    plt.figure()
    sns.histplot(normal_distribuition)
    plt.title(f"Distribuicao Normal com {n} elementos com média {media} e std {desvio_padrao}")
    plt.xlabel("Valor")
    plt.ylabel("Frequencia")
    plt.show()
    print(f"A média é de:{normal_distribuition.mean():.2f} \nDesvio padrao:{normal_distribuition.std():.2f}")
    n*=10
```

```python
n = 100
p = 0.3
tentativas = 3

for i in range(3):    
    binomial_distribuition = np.random.binomial(n=3,p=0.3,size=n)
    plt.figure()
    sns.countplot(x=binomial_distribuition)
    plt.title(f"Distribuicao Binomial com {n} elementos, com {tentativas} tentativas e {p}% de dar certo")
    plt.xlabel("Quantos Sucessos")
    plt.ylabel("Frequencia")
    plt.show()
    print(f"A média é de:{binomial_distribuition.mean():.2f} \nDesvio padrao:{binomial_distribuition.std():.2f}")
    n*=10
```

```python
n = 100
media = 1
for i in range(3):    
    poisson_distribuition = np.random.poisson(lam = media,size = n)
    plt.figure()
    sns.countplot(x=poisson_distribuition)
    plt.title(f"Distribuicao Poisson com {n} elementos e média de evento por intervalo: {media}")
    plt.xlabel("Número de eventos")
    plt.ylabel("Frequência (quantidade)")
    plt.show()
    print(f"A média é de:{poisson_distribuition.mean():.2f} \nDesvio padrao:{poisson_distribuition.std():.2f}")
    n*=10
```
